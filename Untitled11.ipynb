```python
# ================================
# FULL CNN DROPOUT EXPERIMENT
# ================================

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ================================
# DATASET (GTSRB - TRANSPORT)
# ================================

transform = transforms.Compose([
    transforms.Resize((32, 32)),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

trainset = torchvision.datasets.GTSRB(
    root='./data',
    split='train',
    download=True,
    transform=transform
)

testset = torchvision.datasets.GTSRB(
    root='./data',
    split='test',
    download=True,
    transform=transform
)

trainloader = torch.utils.data.DataLoader(trainset, batch_size=64, shuffle=True)
testloader = torch.utils.data.DataLoader(testset, batch_size=64, shuffle=False)

# ================================
# MODELS
# ================================

class CNN_NoDropout(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2,2),
            nn.Conv2d(16, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2,2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32*8*8, 128),
            nn.ReLU(),
            nn.Linear(128, 43)
        )
    def forward(self, x):
        return self.classifier(self.features(x))


class CNN_Dropout(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2,2),
            nn.Conv2d(16, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2,2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32*8*8, 128),
            nn.ReLU(),
            nn.Dropout(0.5),   # KEY DIFFERENCE
            nn.Linear(128, 43)
        )
    def forward(self, x):
        return self.classifier(self.features(x))

# ================================
# TRAIN FUNCTION
# ================================

def train_model(model, trainloader, testloader, epochs=5):
    model.to(device)
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=0.001)

    train_losses = []
    test_losses = []

    for epoch in range(epochs):
        model.train()
        running_loss = 0

        for images, labels in trainloader:
            images, labels = images.to(device), labels.to(device)

            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            running_loss += loss.item()

        train_loss = running_loss / len(trainloader)
        train_losses.append(train_loss)

        # Evaluation
        model.eval()
        test_loss = 0
        with torch.no_grad():
            for images, labels in testloader:
                images, labels = images.to(device), labels.to(device)
                outputs = model(images)
                loss = criterion(outputs, labels)
                test_loss += loss.item()

        test_loss /= len(testloader)
        test_losses.append(test_loss)

        print(f"Epoch {epoch+1}: Train={train_loss:.3f}, Test={test_loss:.3f}")

    return train_losses, test_losses

# ================================
# TRAIN BOTH MODELS
# ================================

model_no = CNN_NoDropout()
model_do = CNN_Dropout()

print("\nTraining WITHOUT Dropout")
loss_no, test_no = train_model(model_no, trainloader, testloader, epochs=3)

print("\nTraining WITH Dropout")
loss_do, test_do = train_model(model_do, trainloader, testloader, epochs=3)

# ================================
# LOSS GRAPH
# ================================

plt.figure()
plt.plot(loss_no, label="Train No Dropout")
plt.plot(test_no, label="Test No Dropout")
plt.plot(loss_do, label="Train Dropout")
plt.plot(test_do, label="Test Dropout")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.title("Dropout vs No Dropout")
plt.show()

# ================================
# KERNEL DISTRIBUTION (NO VISUAL IMAGES)
# ================================

def plot_kernel_distribution(model, title):
    weights = model.features[0].weight.data.cpu().numpy().flatten()
    plt.figure()
    plt.hist(weights, bins=50)
    plt.xlabel("Weight Value")
    plt.ylabel("Frequency")
    plt.title(title)
    plt.show()

plot_kernel_distribution(model_no, "Kernel Distribution - No Dropout")
plot_kernel_distribution(model_do, "Kernel Distribution - Dropout")

# ================================
# KERNEL STATISTICS
# ================================

def kernel_stats(model, name):
    weights = model.features[0].weight.data.cpu().numpy()
    print(f"\n{name}")
    print("Mean:", np.mean(weights))
    print("Std:", np.std(weights))
    print("Max:", np.max(weights))
    print("Min:", np.min(weights))

kernel_stats(model_no, "No Dropout Model Stats")
kernel_stats(model_do, "Dropout Model Stats")
```
